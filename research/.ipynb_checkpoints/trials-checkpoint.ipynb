```python
from dataclasses import dataclass
from pathlib import Path

@dataclass(frozen=True)  # frozen=True giúp object này không bị thay đổi sau khi tạo (immutable)
class DataIngestionConfig:
    root_dir: Path
    source_url: str
    local_data_file: Path
    unzip_dir: Path
```
